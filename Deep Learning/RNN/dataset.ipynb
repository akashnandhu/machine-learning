```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,SimpleRNN,Bidirectional
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=10000)
print("training samples",len(x_train))
print("testing samples",len(x_test))

```

```python
x_train=pad_sequences(x_train,maxlen=200)
x_test=pad_sequences(x_test,maxlen=200)
x_test
```

```python
model=Sequential([
    Embedding(10000,200),
    SimpleRNN(200,activation="relu"),
    Dense(1,activation="sigmoid")
])
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
model.summary()
```

```python
history=model.fit(
    x_train,y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)
```

```python
loss,accuracy=model.evaluate(x_test,y_test,verbose=1)
print(accuracy)
```

```python
word_index=imdb.get_word_index()
def encode_review(text):
    encoded=[word_index.get(word,2) for word in text.lower().split()]
    return pad_sequences([encoded],maxlen=200)
review="The movie was absolutely fantastic and enjoyable"
prediction=model.predict(encode_review(review))

print("Positive" if prediction[0][0] > 0.5 else "Negative")
```
